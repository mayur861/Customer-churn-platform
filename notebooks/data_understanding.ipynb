```python
# Find rows where TotalCharges cannot be converted to numeric
invalid_total = df[pd.to_numeric(df["TotalCharges"], errors="coerce").isna()]

print(invalid_total[["customerID", "tenure", "TotalCharges", "Churn"]])
```

```python
# Convert TotalCharges from object to numeric
df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"],
    errors="coerce"
)

# Zero-tenure customers ke blank TotalCharges ko 0 se fill karo
df["TotalCharges"] = df["TotalCharges"].fillna(0)

# Verification
print("TotalCharges dtype:", df["TotalCharges"].dtype)
print("Missing TotalCharges:", df["TotalCharges"].isnull().sum())
print("Rows:", df.shape[0])
```
